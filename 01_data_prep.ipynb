```python
import pandas as pd

# Load only the columns you need from each file
econ_cols = ['GeoID', 'GeogName', 'Borough', 'MdHHIncE', 'PerCapIncE', 
             'CW_PbTrnsP', 'CW_DrvAlnP', 'PBwPvP', 'MgBSciArtP']

soc_cols = ['GeoID', 'EA_BchDHP', 'EA_LTHSGrP', 'AvgHHSzE', 'Fb1P']

econ = pd.read_excel('Econ_1923_NTA.xlsx', usecols=econ_cols)
soc = pd.read_excel('Soc_1923_NTA.xlsx', usecols=soc_cols)

# Merge them together on NTA code
demographics = econ.merge(soc, on='GeoID')

print(demographics.shape)
demographics.head()
```

```python
taxi_lookup = pd.read_csv('taxi_zone_lookup.csv')
print(taxi_lookup.shape)
print(taxi_lookup['Zone'].tolist())

print(demographics['GeogName'].tolist())
```

```python
import geopandas as gpd
taxi_zones = gpd.read_file('taxi_zones/taxi_zones.shp')
print(taxi_zones.columns.tolist())
taxi_zones.head()
```

```python
nta = gpd.read_file('nta_zones/geo_export.shp')
print(nta.columns.tolist())
nta.head()
```

```python
# Make sure both are in the same coordinate system
taxi_zones = taxi_zones.to_crs(nta.crs)

# Spatial join - match each TLC zone to the NTA it overlaps with most
crosswalk = gpd.sjoin(taxi_zones, nta[['nta2020', 'ntaname', 'geometry']], 
                      how='left', 
                      predicate='intersects')

# Keep only the columns you need
crosswalk = crosswalk[['LocationID', 'zone', 'borough', 'nta2020', 'ntaname']]

print(crosswalk.shape)
crosswalk.head(10)
```

```python
import shapely

# Reproject both to a projected CRS so area calculations are accurate
taxi_proj = taxi_zones.to_crs(epsg=2263)
nta_proj = nta.to_crs(epsg=2263)

# Spatial join
joined = gpd.sjoin(taxi_proj, nta_proj[['nta2020', 'ntaname', 'geometry']], 
                   how='left', predicate='intersects')

# Calculate actual intersection area between each matched pair
joined['overlap_area'] = joined.apply(
    lambda row: taxi_proj.loc[taxi_proj.LocationID == row.LocationID, 'geometry'].values[0]
    .intersection(nta_proj.loc[nta_proj.nta2020 == row.nta2020, 'geometry'].values[0])
    .area if pd.notna(row.nta2020) else 0, axis=1
)

# Keep only the best match per LocationID
crosswalk_clean = (joined.sort_values('overlap_area', ascending=False)
                         .drop_duplicates(subset='LocationID', keep='first')
                         [['LocationID', 'zone', 'borough', 'nta2020', 'ntaname']]
                         .reset_index(drop=True))

print(crosswalk_clean.shape)
crosswalk_clean.head(10)
```

```python
# Merge crosswalk with demographics
zone_demographics = crosswalk_clean.merge(demographics, 
                                           left_on='nta2020', 
                                           right_on='GeoID', 
                                           how='left')

print(zone_demographics.shape)
print(zone_demographics.isnull().sum())
zone_demographics.head()
```

```python
zone_demographics.to_csv('zone_demographics.csv', index=False)
```

```python
import duckdb

query = """
    SELECT 
        t.PULocationID,
        t.hvfhs_license_num,
        COUNT(*) as total_trips,
        AVG(t.tips) as avg_tip,
        AVG(t.tips / NULLIF(t.base_passenger_fare, 0)) as avg_tip_rate,
        SUM(CASE WHEN t.tips > 0 THEN 1 ELSE 0 END) * 100.0 / COUNT(*) as pct_tipped
    FROM 'fhvhv_tripdata_2024-*.parquet' t
    WHERE t.tips IS NOT NULL
    GROUP BY t.PULocationID, t.hvfhs_license_num
"""

zone_trips = duckdb.query(query).to_df()
```

```python
final = zone_trips.merge(zone_demographics, 
                          left_on='PULocationID', 
                          right_on='LocationID',
                          how='left')
print(final.shape)
final.columns.tolist()
```

```python
final = final.rename(columns={
    'hvfhs_license_num': 'service',
    'CW_DrvAlnP': 'pct_drive_alone',
    'CW_PbTrnsP': 'pct_public_transit',
    'MgBSciArtP': 'pct_white_collar',
    'MdHHIncE': 'median_hh_income',
    'PerCapIncE': 'per_capita_income',
    'PBwPvP': 'pct_poverty',
    'AvgHHSzE': 'avg_household_size',
    'EA_LTHSGrP': 'pct_no_hs_diploma',
    'EA_BchDHP': 'pct_bachelors_plus',
    'Fb1P': 'pct_foreign_born',
    'GeogName': 'neighborhood_name',
    'GeoID': 'nta_code'
})

# Also replace service codes with readable names
final['service'] = final['service'].map({
    'HV0003': 'Uber',
    'HV0005': 'Lyft'
})

print(final.columns.tolist())
final.head(10)
```

```python
final.sort_values(['PULocationID', 'service'])[
    ['PULocationID', 'neighborhood_name', 'service', 
     'total_trips', 'avg_tip', 'avg_tip_rate', 
     'pct_tipped', 'median_hh_income', 'pct_bachelors_plus']
].reset_index(drop=True)
```

```python
print(final['median_hh_income'].isna().sum(), 'zones missing demographics out of', len(final))
print(final[final['median_hh_income'].isna()][['PULocationID', 'neighborhood_name', 'service', 'total_trips']].drop_duplicates('PULocationID'))

# Check if NaN pattern is the same across all demographic columns
print(final['median_hh_income'].isna().sum())
print(final['pct_bachelors_plus'].isna().sum())
print(final['pct_poverty'].isna().sum())
print(final['pct_public_transit'].isna().sum())
```

```python
model_vars = ['median_hh_income', 'pct_bachelors_plus', 'pct_poverty', 
              'pct_public_transit', 'pct_white_collar', 'per_capita_income',
              'pct_no_hs_diploma', 'pct_foreign_born']

final_clean = final.dropna(subset=model_vars)
print(final_clean.shape)
final_clean.to_csv('final_clean.csv', index=False)
```
